```python

# Configuração de Namespaces
NAMESPACES = {
    'interface': "http://www.siemens.com/automation/Openness/SW/Interface/v5",
    'flgnet': "http://www.siemens.com/automation/Openness/SW/NetworkSource/FlgNet/v4"
}

def create_tia_xml():
    # Raiz do documento
    root = ET.Element("Document")
    
    # Engineering
    ET.SubElement(root, "Engineering", version="V17")
    
    # DocumentInfo
    doc_info = ET.SubElement(root, "DocumentInfo")
    ET.SubElement(doc_info, "Created").text = datetime.now().isoformat() + "Z"
    ET.SubElement(doc_info, "ExportSetting").text = "None"
    
    # Produtos Instalados
    installed_products = ET.SubElement(doc_info, "InstalledProducts")
    products = [
        ("Product", "Totally Integrated Automation Portal", "V17"),
        ("OptionPackage", "TIA Portal Openness", "V17"),
        ("OptionPackage", "TIA Portal Version Control Interface", "V17"),
        ("Product", "STEP 7 Professional", "V17"),
        ("OptionPackage", "STEP 7 Safety", "V17"),
        ("Product", "WinCC Professional", "V17"),
    ]
    
    for tag, name, version in products:
        item = ET.SubElement(installed_products, tag)
        ET.SubElement(item, "DisplayName").text = name
        ET.SubElement(item, "DisplayVersion").text = version

    # Bloco SW.Blocks.FC
    fc_block = ET.SubElement(root, "SW.Blocks.FC", ID="0")
    attr_list = ET.SubElement(fc_block, "AttributeList")
    
    # Interface (com Namespace)
    interface = ET.SubElement(attr_list, "Interface")
    sections = ET.SubElement(interface, "Sections", xmlns=NAMESPACES['interface'])
    for sec_name in ["Input", "Output", "InOut", "Temp", "Constant"]:
        ET.SubElement(sections, "Section", Name=sec_name)
    
    ret_section = ET.SubElement(sections, "Section", Name="Return")
    ET.SubElement(ret_section, "Member", Name="Ret_Val", Datatype="Void")

    # Atributos Adicionais
    ET.SubElement(attr_list, "MemoryLayout").text = "Optimized"
    ET.SubElement(attr_list, "Name").text = "Block_1"
    ET.SubElement(attr_list, "Number").text = "1"
    ET.SubElement(attr_list, "ProgrammingLanguage").text = "LAD"
    ET.SubElement(attr_list, "SetENOAutomatically").text = "false"

    # ObjectList
    obj_list = ET.SubElement(fc_block, "ObjectList")
    
    # Comentário Multilíngue
    multi_text = ET.SubElement(obj_list, "MultilingualText", ID="1", CompositionName="Comment")
    multi_obj_list = ET.SubElement(multi_text, "ObjectList")
    multi_item = ET.SubElement(multi_obj_list, "MultilingualTextItem", ID="2", CompositionName="Items")
    item_attr = ET.SubElement(multi_item, "AttributeList")
    ET.SubElement(item_attr, "Culture").text = "en-US"
    ET.SubElement(item_attr, "Text")

    # Unidade de Compilação (Network)
    compile_unit = ET.SubElement(obj_list, "SW.Blocks.CompileUnit", ID="3", CompositionName="CompileUnits")
    comp_attr = ET.SubElement(compile_unit, "AttributeList")
    ET.SubElement(comp_attr, "NetworkSource")
    # Nota: Para adicionar conteúdo ao FlgNet, use o namespace NAMESPACES['flgnet']

    # Gerar a árvore e salvar
    tree = ET.ElementTree(root)
    ET.indent(tree, space="  ", level=0) # Formatação amigável (Python 3.9+)
    
    with open("output.xml", "wb") as f:
        f.write(b'<?xml version="1.0" encoding="utf-8"?>\n')
        tree.write(f, encoding="utf-8", xml_declaration=False)

if __name__ == "__main__":
    create_tia_xml()
    print("XML gerado com sucesso!")
```

```python
#Bloco criação do elemento no texto
lista_uid = [] 
lista_dicionarios =[]
lista_nomear = []# lista com os nomes de cada elemento
lista_de = [] #lista com a declaracao em ingles de cada elemento
dicionario_elementos = {"bobina":"coil","contato":"Contact"}
dicionario_nomear_temporario ={}
dicionario_nomear={}
dicionario_nomear_invertido = {}
dicionario_casal_elementos ={}
lista_dicionarios_invertido = {}
dicionario_casal_invertido = {}
dicionario_nome_elemento = {}
lista_esquerda =[]
lista_uid.append(20)
def nomear_elemento(nome, tipo,extra= 1 ):
    lista_uid.append(lista_uid[-1]+1)
    access = ET.Element("Access", Scope ="GlobalVariable", UId = str(lista_uid[-1]) )
    symbol = ET.SubElement(access, "Symbol")
    ET.SubElement(symbol, "Component", Name=str(nome))
    ET.indent(access, space="    ", level=0)
    xml_string = ET.tostring(access, encoding='unicode')
    dicionario_nomear_temporario ={str(lista_uid[-1]):str(nome)}
    dicionario_nomear.update(dicionario_nomear_temporario)#criando um dicionario com um nome e a UId do nome que eu criei
    dicionario_nomear_temporario = {valor: chave for chave, valor in dicionario_nomear_temporario.items()}
    dicionario_nomear_invertido.update(dicionario_nomear_temporario)
    print(xml_string)
    lista_nomear.append(xml_string)
    return(dicionario_nomear)

def definir_elemento(nome, tipo):
    lista_uid.append(lista_uid[-1]+1)
    part = ET.Element("Part", Name = dicionario_elementos[tipo], UId = str(lista_uid[-1]))
    xml_string_2 = ET.tostring(part, encoding='unicode')
    #lista definir
    lista_de.append(xml_string_2)
    dicionario_casal_temporario = {str(lista_uid[-1]):str(lista_uid[-2])}
    dicionario_casal_elementos.update(dicionario_casal_temporario)
    dicionario_casal_temporario = {valor: chave for chave, valor in dicionario_casal_temporario.items()}
    dicionario_casal_invertido.update(dicionario_casal_temporario)
    tipo = tipo
    print(xml_string_2)

def elemento_completo(tipo):
    tipo = tipo
    ultima_chave_nome = next(reversed(dicionario_nomear))
    ultimo_valor_nome = dicionario_nomear[ultima_chave_nome]
    ultima_chave_elemento = next(reversed(dicionario_casal_elementos))
    ultimo_valor_elemento = dicionario_casal_elementos[ultima_chave_elemento]
    dicionario_nome_elemento.update({(ultimo_valor_nome,tipo):[ultima_chave_nome,ultima_chave_elemento]})

def parte_esquerda(obj1):
    #Cria mais um #part referenciando ao objeto O de adição de elementos em um mesmo no
    lista_uid.append(lista_uid[-1]+1)
    no_conexao = lista_uid[-1]
    part = ET.Element("Part", Name="O", UId=str(lista_uid[-1]))
    template_value = ET.SubElement(part, "TemplateValue", Name="Card", Type="Cardinality")
    template_value.text = str(len(obj1))
    ET.indent(part, space="    ", level=0)
    xml_string = ET.tostring(part, encoding='unicode')
    lista_esquerda.append(xml_string)
    return(no_conexao)

def criar_elemento(nome,tipo,extra):
    nomear_elemento(nome,tipo,1)
    definir_elemento(nome,tipo)
    elemento_completo(tipo)

```

```python
#FIOS
#Ate agora so falei sobre os elementos
#Fios sao divididos em 4 partes: geralzao, dados input, dados output, conexao.
#geralzao
lista_obj = []
lista_wires = []
contador_wires = 0
#contador_wires = contador_wires +1
#geralzao
def fio_geral(obj1,obj2,n_or = 1):
    lista_wires.append([])
    lista_uid.append(lista_uid[-1]+1)
    #tipo = dicionario_casal_invertido[obj1]
    #no base que estara ligado ao power rail
    wire = ET.Element("Wire", UId = str(lista_uid[-1]))
    ET.SubElement(wire,"Powerrail")
    ET.SubElement(wire,"NameCon", UId= dicionario_nome_elemento[obj1][1] , Name="in")
    ET.indent(wire, space="  ")
    xml_output_w1 = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_output_w1)

    
def fio_geral_varios(obj1,obj2,n_or =1):
    #Funcao ligada ao powerrail. No futuro vou deixala como opcional para ser ativada ou não
    lista_wires.append([])
    lista_uid.append(lista_uid[-1]+1)
    #Mesma coisa do fio geral, agora para varias entradas de input
    wire = ET.Element("Wire", UId = str(lista_uid[-1]))
    ET.SubElement(wire,"Powerrail")
    for x in range(len(obj1)):
        ET.SubElement(wire,"NameCon", UId= dicionario_nome_elemento[obj1[x]][1] , Name="in")
    ET.indent(wire, space="  ")
    xml_output_w1 = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_output_w1)


 

def fiod_input(obj1,):#n_or numero de or no mesmo bloco
    #faz a relação nome material do elemento
    lista_uid.append(lista_uid[-1]+1)
    #Mesmo passo para os outros    
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "IdentCon", UId=dicionario_nome_elemento[obj1][0])
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj1][1], Name="operand")
    ET.indent(wire, space="  ")
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)




def fiod_input_varios(obj1,no_conexao):
    #Cria o input para varios elementos. Inicialmente estou contando a esquerda
    lista_uid.append(lista_uid[-1]+1)
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "IdentCon", UId=dicionario_nome_elemento[obj1][0])
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj1][1], Name="operand")
    ET.indent(wire, space="  ")
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)

         
         



def fiod_conect(obj1,obj2,n_or = 1):#Conectar dois elementos(codigo dos elementos)    
    lista_uid.append(lista_uid[-1]+1)
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj1][1],Name="out")
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj2][1], Name="in")
    ET.indent(wire, space="  ")
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)
    


def fiod_conect_varios(obj1,obj2,contador_fio_master,n_or=1):#Conectar varios elementos a um no
    '''
    obj1 é a minha lista que vou trabalhar
    obj2 é o numero do meu nó que vai conectar todos os outros elementos
    contador_fio_master é a posição da minha lista do fio master que vou receber
    n_or ate agora foi um dummy, mas vou tiralo no futuro
    '''
    lista_uid.append(lista_uid[-1]+1)
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "NameCon", UId=str(dicionario_nome_elemento[obj1[contador_fio_master]][1]),Name="out")
    ET.SubElement(wire, "NameCon", UId=str(obj2), Name="in"+str(contador_fio_master+1))
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)


def fiod_final(obj1,obj2):
    #identificar o elemento que esta no outro no, definindo que a bobina 1 tem nome relacionado a variavel x ao tipo de componente da bobina 2
    lista_uid.append(lista_uid[-1]+1)
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "IdentCon", UId=dicionario_nome_elemento[obj2][0])
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj2][1], Name="operand")
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)

def conexao_no_esquerdo_fio(no_fios,obj2):
    #no_fio é o UId do no do fio
    '''É a função que faz a conexão do nó esquerdo com o fio ligado a ela na direita
    -------x--------()
           | 
    --------
           |
    --------'''
    '''No desenho acima ele faz a conexao do meu nó(ponto x) com a minha bobina(). Essa conexao pode ser no com no, no com fio, etc'''
    lista_uid.append(lista_uid[-1]+1)
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "NameCon", UId=str(no_fios),Name="out")
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj2][1], Name="in")
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)

def no_sem_card(obj1,obj2):
    '''Faz as conexoes out/in para varios fios. Como não existe contato  em paralelo do outro lado
    eu posso somente fazer as conexoes dos fios'''
    lista_uid.append(lista_uid[-1]+1)
    wire = ET.Element("Wire", UId = str(lista_uid[-1]) )
    ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj1][1], Name="out")
    for x in range(len(obj2)):
        ET.SubElement(wire, "NameCon", UId=dicionario_nome_elemento[obj2[x]][1], Name="in")
    xml_string = ET.tostring(wire, encoding="unicode")
    lista_wires[contador_wires].append(xml_string)




    




#def fiod_final_varios(obj1,obj2,n_or):



def fio_master(obj1,obj2,n_or = 1):
    #fio master gera obrigatoriamente uma conexao com o powerrail. Ainda vou encontrar outra maneira de emendar fios
    #obj1 elementos da esquerda
    #obj2 elementos da direita
    '''n_or eh o numero de or dentro de um elemento.
      Se o usuario não fornecer valor para n or, o valor assumido sera obrigatoriamente 1'''
    '''Ate o momento, a quantidade de fios em paralelo obrigatoriamente deve fser referenciada a obj1
    em uma lista que vai se juntar obrigatoriamente ao contato/bobina'''
    if type(obj1) == tuple and type(obj2 )== tuple:
        print('modo1')
        fio_geral(obj1,obj2)
        fiod_input(obj1)
        fiod_conect(obj1,obj2)
        fiod_final(obj1,obj2)
    elif type(obj1) == list and type(obj2) == tuple:
        print('modo 2')
        no_conexao = parte_esquerda(obj1)
        fio_geral_varios(obj1,1,1)
        for x in range(len(obj1)):
            fiod_input_varios(obj1[x],no_conexao)
            fiod_conect_varios(obj1,no_conexao,x,1)
        conexao_no_esquerdo_fio(no_conexao,obj2)
        fiod_final(obj1,obj2)
    elif type(obj1) == tuple and type(obj2) == list:
        print('modo3')
        fio_geral(obj1,obj2)
        fiod_final(obj2,obj1)
        no_sem_card(obj1,obj2)
        for x in range(len(obj2)):
            fiod_final(obj1,obj2[x])
    else:
        print('Formato de Entrada Invalido')

    
def existe_na_lista(elemento, lista):
    for item in lista:
        # Se o item for uma lista, busca recursivamente nela
        if isinstance(item, list):
            if existe_na_lista(elemento, item):
                return True
    



def adicionar_direita(obj1,obj2,lista_obj = lista_obj):
    '''Tenho o obj1 e quero adicionar um elemento obj2 a sua direita. O elemento obj2
     O elemento obj2 so vai ser adicionado se nao tiver NADA a direita de obj1 '''
    if existe_na_lista(obj1, lista_obj):
        print('Refaça o codigo, ha um elemento depois dele')
    else:
        #Dessa vez nao e necessario adicionar nada ao powerrail
        fiod_conect(obj1,obj2)
        fiod_final(obj1,obj2)

def adicionar_direita_varios(obj1,obj2,lista_obj = lista_obj):
    '''Tenho o obj1 e quero adicionar um elemento obj2 a sua direita. O elemento obj2
     O elemento obj2 so vai ser adicionado se nao tiver NADA a direita de obj1 '''
    if existe_na_lista(obj1, lista_obj):
        print('Refaça o codigo, ha um elemento depois dele')
    else:
        #Novamente nao e necessario adicionar nada ao powerrail
        no_sem_card(obj1,obj2)
        for x in range(len(obj2)):
            fiod_final(obj1,obj2[x])


    
    
```
