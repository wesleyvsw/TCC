```python
def diferenciar_nomes(dados_ladder):
    contador = {}
    for x in dados_ladder:
        elemento = x[0]
        if elemento in contador:
            contador[elemento] +=1
        else:
            contador[elemento] = 1
        repeticoes = contador[elemento] -1
        x[0] = x[0][:-1] + (1*repeticoes,)
    return(dados_ladder)
```

```python
def ordenar_tipo_elemento(dados):
    #Com a lista ja ordenada, agora vou ordenar o elemento para colocalo na part do meu xml
    contagem = Counter([x[3] for x in dados])
    vistos = {}
    resultado = []

    for item in dados:
        valor = item[3]

        # conta quantas vezes já vi esse valor
        vistos[valor] = vistos.get(valor, 0) + 1

        resultado.append(item)

        # se for a ÚLTIMA ocorrência e tiver repetição
        if contagem[valor] > 1 and vistos[valor] == contagem[valor]:
            resultado.append(['card', contagem[valor]])

    return resultado
```

```python
def gerar_chaves_ladder(lista_elementos):
    elementos_dict = {}
    
    # 1. Registrar todos os elementos (a ordem da lista dita a prioridade "cima para baixo")
    for i, el in enumerate(lista_elementos):
        elementos_dict[i] = {
            "nome": el[0], "proximo": el[1], 
            "no_in": el[2], "no_out": el[3],
            "predecessores": set(), "sucessores": set()
        }
        # 2. Mapear conexões dos nós - cada no_out- no de saida de uma elemento vai ganhar o numero de seus elementos que saem dele
    saidas_por_no = {}
    for i, el in elementos_dict.items():
        if el["no_out"] not in saidas_por_no:
            saidas_por_no[el["no_out"]] = []
        saidas_por_no[el["no_out"]].append(i)

    for i, el in elementos_dict.items():
        if el["no_in"] in saidas_por_no:
            for pred_id in saidas_por_no[el["no_in"]]:
                el["predecessores"].add(pred_id)
                elementos_dict[pred_id]["sucessores"].add(i)
    for i, el in elementos_dict.items():
        if el["no_in"] in saidas_por_no:
            for pred_id in saidas_por_no[el["no_in"]]:
                el["predecessores"].add(pred_id)
                elementos_dict[pred_id]["sucessores"].add(i)
    
    # 4. Ordenação Topológica com prioridade de índice
    prontos = [i for i, el in elementos_dict.items() if len(el["predecessores"]) == 0]
    grau_entrada = {i: len(el["predecessores"]) for i, el in elementos_dict.items()}
    
    resultado = []
    contador = 1
    
    while prontos:
        prontos.sort() 
        
        atual_id = prontos.pop(0)
        atual = elementos_dict[atual_id]
        
        resultado.append({
            "Chave": contador,
            "Elemento": atual["nome"]
        })
        contador += 1
        
        for suc_id in atual["sucessores"]:
            grau_entrada[suc_id] -= 1
            if grau_entrada[suc_id] == 0:
                prontos.append(suc_id)
                
    return resultado
```

```python
def gerar_chaves_ladder_robusto(lista_elementos):
    elementos_dict = {}
    for i, el in enumerate(lista_elementos):
        elementos_dict[i] = {
            "nome": el[0], "no_in": el[2], "no_out": el[3],
            "predecessores": set(), "sucessores": set(),
            "indice_original": i,
            "y_score": None # A "Linha Virtual" que vamos calcular
        }

    # 1. Construir Conexões (Igual antes)
    saidas_por_no = {}
    for i, el in elementos_dict.items():
        if el["no_out"] not in saidas_por_no: saidas_por_no[el["no_out"]] = []
        saidas_por_no[el["no_out"]].append(i)

    for i, el in elementos_dict.items():
        if el["no_in"] in saidas_por_no:
            for pred_id in saidas_por_no[el["no_in"]]:
                el["predecessores"].add(pred_id)
                elementos_dict[pred_id]["sucessores"].add(i)

    # ==========================================
    # NOVIDADE: PASSO 1 - CALCULAR A "LINHA VIRTUAL" (Y-SCORE)
    # ==========================================
    # Pega quem está ligado direto na barra (Nó 0)
    raizes = [i for i, el in elementos_dict.items() if len(el["predecessores"]) == 0]
    raizes.sort(key=lambda x: elementos_dict[x]["indice_original"])
    
    # Dá a eles uma "linha" base (0, 1, 2, 3...)
    for linha, id_raiz in enumerate(raizes):
        elementos_dict[id_raiz]["y_score"] = linha

    # Propaga essa "linha" para frente no circuito
    grau_entrada_y = {i: len(el["predecessores"]) for i, el in elementos_dict.items()}
    fila_y = raizes[:]
    
    while fila_y:
        atual_id = fila_y.pop(0)
        atual = elementos_dict[atual_id]
        
        for suc_id in atual["sucessores"]:
            suc = elementos_dict[suc_id]
            
            # O sucessor herda a linha. Se vários se juntam, ele fica com a mais alta (menor valor)
            if suc["y_score"] is None:
                suc["y_score"] = atual["y_score"]
            else:
                suc["y_score"] = min(suc["y_score"], atual["y_score"])
            
            grau_entrada_y[suc_id] -= 1
            if grau_entrada_y[suc_id] == 0:
                fila_y.append(suc_id)

    # ==========================================
    # PASSO 2 - NUMERAÇÃO OFICIAL (A CHAVE VERMELHA)
    # ==========================================
    grau_entrada = {i: len(el["predecessores"]) for i, el in elementos_dict.items()}
    prontos = [i for i, el in elementos_dict.items() if grau_entrada[i] == 0]
    
    resultado = []
    contador = 1
    
    while prontos:
        # A MÁGICA ACONTECE AQUI:
        # Ordena PRIMEIRO pela Linha Virtual (y_score), e só usa o índice original como desempate final
        prontos.sort(key=lambda x: (elementos_dict[x]["y_score"], elementos_dict[x]["indice_original"]))
        
        atual_id = prontos.pop(0)
        atual = elementos_dict[atual_id]
        
        resultado.append({
            "Chave": contador, 
            "Elemento": atual["nome"], 
            "Linha_Virtual": atual["y_score"] # Apenas para você ver no print
        })
        contador += 1
        
        for suc_id in atual["sucessores"]:
            grau_entrada[suc_id] -= 1
            if grau_entrada[suc_id] == 0:
                prontos.append(suc_id)
                
    return resultado

# TESTE: Vou colocar o cont5 no final da lista de propósito!
dados_baguncados = [
    ["cont1", "cont9_sup", 0, 1],
    ["cont3", "cont5", 0, 2],
    ["cont4", "cont5", 0, 2],
    ["cont9_inf", "cont10", 0, 4],
    ["cont9_sup", "cont5", 1, 2],
    ["cont10", "nó_central", 4, 3],
    ["cont6", "bobina1", 3, 5],
    ["bobina1", "fim_linha1", 5, 6],
    ["cont7", "cont8", 3, 7],
    ["cont8", "bobina2", 7, 8],
    ["bobina2", "fim_linha2", 8, 9],
    ["cont5", "nó_central", 2, 3]  # <--- COLOCADO NO FINAL PARA TESTAR O SEU CASO
]

for item in gerar_chaves_ladder_robusto(dados_baguncados):
    print(f"Chave: {item['Chave']:02d} | Elemento: {item['Elemento']:<10} | Herdou Linha: {item['Linha_Virtual']}")
```
