```python

# Configuração de Namespaces
NAMESPACES = {
    'interface': "http://www.siemens.com/automation/Openness/SW/Interface/v5",
    'flgnet': "http://www.siemens.com/automation/Openness/SW/NetworkSource/FlgNet/v4"
}

def create_tia_xml():
    # Raiz do documento
    root = ET.Element("Document")
    
    # Engineering
    ET.SubElement(root, "Engineering", version="V17")
    
    # DocumentInfo
    doc_info = ET.SubElement(root, "DocumentInfo")
    ET.SubElement(doc_info, "Created").text = datetime.now().isoformat() + "Z"
    ET.SubElement(doc_info, "ExportSetting").text = "None"
    
    # Produtos Instalados
    installed_products = ET.SubElement(doc_info, "InstalledProducts")
    products = [
        ("Product", "Totally Integrated Automation Portal", "V17"),
        ("OptionPackage", "TIA Portal Openness", "V17"),
        ("OptionPackage", "TIA Portal Version Control Interface", "V17"),
        ("Product", "STEP 7 Professional", "V17"),
        ("OptionPackage", "STEP 7 Safety", "V17"),
        ("Product", "WinCC Professional", "V17"),
    ]
    
    for tag, name, version in products:
        item = ET.SubElement(installed_products, tag)
        ET.SubElement(item, "DisplayName").text = name
        ET.SubElement(item, "DisplayVersion").text = version

    # Bloco SW.Blocks.FC
    fc_block = ET.SubElement(root, "SW.Blocks.FC", ID="0")
    attr_list = ET.SubElement(fc_block, "AttributeList")
    
    # Interface (com Namespace)
    interface = ET.SubElement(attr_list, "Interface")
    sections = ET.SubElement(interface, "Sections", xmlns=NAMESPACES['interface'])
    for sec_name in ["Input", "Output", "InOut", "Temp", "Constant"]:
        ET.SubElement(sections, "Section", Name=sec_name)
    
    ret_section = ET.SubElement(sections, "Section", Name="Return")
    ET.SubElement(ret_section, "Member", Name="Ret_Val", Datatype="Void")

    # Atributos Adicionais
    ET.SubElement(attr_list, "MemoryLayout").text = "Optimized"
    ET.SubElement(attr_list, "Name").text = "Block_1"
    ET.SubElement(attr_list, "Number").text = "1"
    ET.SubElement(attr_list, "ProgrammingLanguage").text = "LAD"
    ET.SubElement(attr_list, "SetENOAutomatically").text = "false"

    # ObjectList
    obj_list = ET.SubElement(fc_block, "ObjectList")
    
    # Comentário Multilíngue
    multi_text = ET.SubElement(obj_list, "MultilingualText", ID="1", CompositionName="Comment")
    multi_obj_list = ET.SubElement(multi_text, "ObjectList")
    multi_item = ET.SubElement(multi_obj_list, "MultilingualTextItem", ID="2", CompositionName="Items")
    item_attr = ET.SubElement(multi_item, "AttributeList")
    ET.SubElement(item_attr, "Culture").text = "en-US"
    ET.SubElement(item_attr, "Text")

    # Unidade de Compilação (Network)
    compile_unit = ET.SubElement(obj_list, "SW.Blocks.CompileUnit", ID="3", CompositionName="CompileUnits")
    comp_attr = ET.SubElement(compile_unit, "AttributeList")
    ET.SubElement(comp_attr, "NetworkSource")
    # Nota: Para adicionar conteúdo ao FlgNet, use o namespace NAMESPACES['flgnet']

    # Gerar a árvore e salvar
    tree = ET.ElementTree(root)
    ET.indent(tree, space="  ", level=0) # Formatação amigável (Python 3.9+)
    
    with open("output.xml", "wb") as f:
        f.write(b'<?xml version="1.0" encoding="utf-8"?>\n')
        tree.write(f, encoding="utf-8", xml_declaration=False)

if __name__ == "__main__":
    create_tia_xml()
    print("XML gerado com sucesso!")
```
